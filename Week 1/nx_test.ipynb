```python
import sys
import networkx as nx
import matplotlib
import matplotlib.pyplot as plt

print("Python    :", sys.version.split()[0])
print("NetworkX  :", nx.__version__)
print("matplotlib:", matplotlib.__version__)
```

```python
# Byg en lille graf
G = nx.Graph()
G.add_edges_from([
    ("Anna", "Bo"),
    ("Anna", "Cille"),
    ("Bo", "Cille"),
    ("Cille", "David"),
    ("David", "Eva"),
    ("Eva", "Anna"),
])

print("Knuder:", G.number_of_nodes(), "| Kanter:", G.number_of_edges())
print("Grad pr. knude :", dict(G.degree()))
print("Korteste vej Bo -> Eva :", nx.shortest_path(G, "Bo", "Eva"))
print("Betweenness    :", {k: round(v, 3) for k, v in nx.betweenness_centrality(G).items()})
```

```python
# Klassisk indbygget testgraf
K = nx.karate_club_graph()
print("Zachary karate club:", K.number_of_nodes(), "knuder,", K.number_of_edges(), "kanter")
```

```python
# Tegn grafen
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="#8ecae6", node_size=1500, font_size=10, edge_color="#555")
plt.title("Testgraf – NetworkX virker!")
plt.show()
print("Alt virker ✔")
```
